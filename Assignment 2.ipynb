```python
#WAP to find the median of datapoints without using functions
med=[32,34,35,35,36,37,38,39,39,47]
n=len(med)
```

```python
pos = n // 2
if n % 2 == 0:
    print("Median : ", (med[pos - 1] + med[pos]) / 2)
else:
    print("Median : ", med[pos])
```

```python
# WAP to find a median of datapoints using functions
print("Median is : ",s.median(med))
```

```python
#WAP to find the mode of a data item without using functions
mode1=[1,3,3,3,2,1,2,5,6]
```

```python
frequencies={}
for item in mode1:
    if item in frequencies:
        frequencies[item]+=1
    else:
        frequencies[item]=1

max_frequency = max(frequencies.values())
modes = [key for key, value in frequencies.items() if value == max_frequency]
if max_frequency==1:
    print("No mode found")
else:
    print("Mode :", modes)
```

```python
#WAP to find the mode of a data item with using functions
print("Mode : ",s.mode(mode1))
```

```python
#WAP to print the varience of data points without function
list1=[3,5,5,8,10]
n=len(list1)
x=sum(list1)/n
var=0
for i in range(n):
  var+=(list1[i]-x)**2
vari=var/n
print("Varience : ",vari)
```

```python
#WAP to print the varience of datapoints with functions
print("Variance : ",s.variance(list1))
```

```python
#WAP to print the standred deviation with and without using functions
#Without using library
sd=vari**0.5
print("Standred deviation : ",sd)
print("Standred deviation : ",s.stdev(list1))
```

```python
#WAP to find the variance and standred deviation of the datapoint entered by the user
#Withut using library
n=int(input("Enter a number : "))
list2=[]
for i in range(n):
  list2.append(int(input("Enter a number : ")))
x=0
for i in range(n):
  x+=list2[i]
x=x/n
var=0
for i in range(n):
  var+=(list2[i]-x)**2
vari=var/n
print("Varience : ",vari)
sd=vari**0.5
print("Standred deviation : ",sd)
```
