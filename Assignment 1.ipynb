```python
#WAP to swap two numbers without using third variable
a=int(input("Enter a number : "))
b=int(input("Enter a number : "))
print(f"Before swapping value of a : {a} and b : {b}")
a=a^b
b=a^b
a=a^b
print(f"After swapping value of a : {a} and b : {b}")
```

```python
#WAP to check weather a string is palindrome or not

a=input("Enter a string : ")
n=len(a)
rev=""
for i in range(n-1,-1,-1):
  rev+=a[i]
print("Reverse string is : ",rev)
if(a==rev):
  print(f"{a} is palindrome")
else:
  print(f"{a} is not a palindrome")
```

```python
#WAP to implement a simple calculator
a=float(input("Enter a number : "))
b=float(input("Enter a number : "))
sign=input("Enter operation(+,-,*,/) : ")
if(sign=='+'):
  print("Addition is : ",a+b)
elif(sign=='-'):
  print("Subtraction is : ",a-b)
elif(sign == '*'):
  print("Multiplication is : ",a*b)
elif(sign == '/'):
  try:
    print("Division is : ",a/b)
  except ZeroDivisionError as e:
    print(e)
else:
  print("Invalid sign!")

```

```python
#WAP to count vowels in a string

a=input("Enter a string : ")
cnt=0
for i in a:
  if(i in 'aeiouAEIOU'):
    cnt+=1
print(f"Vowels in string '{a}' are : {cnt}")
```

```python
#WAP to check weather a number is prime or not

a=int(input("Enter a number : "))
cnt=0
for i in range(1,a+1):
  if(a%1==0):
    cnt+=1
if(cnt==2):
  print("Number is prime")
else:
  print("Number is not prime")
```

```python
#WAP to find the sum of digits of a number

a=int(input("Enter a number : "))
sum=0
while(a!=0):
  digit=a%10
  sum+=digit
  a=a//10
print(f"Sum is : {sum}")
```

```python
#WAP to reverse a number

a=int(input("Enter a number : "))
rev=0
while(a!=0):
  digit=a%10
  rev=(rev*10)+digit
  a=a//10
print(f"Reverse is : ",rev)
```

```python
#WAP to find factorial of a number

a=int(input("Enter a number : "))
fact=1
for i in range(1,a+1):
  fact*=i
print(f"Factorial of {a} is : ",fact)
```
