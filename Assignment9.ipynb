# AIM: Associate rule mining
## Dataset1  
Transaction ID |    Items  
T1 |                Bread, Milk  
T2 | Bread,Diaper,Beer,Eggs  
T3 | Milk,Diaper,Beer,Coke  
T4 | Bread,Milk,Diaper,Beer  
T5 | Bread,Milk,Diaper,Coke

Minimum support is 60%  

```python

import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
from scipy.stats import chi2_contingency
import matplotlib.pyplot as plt
import seaborn as sns
import networkx as nx
import warnings
warnings.filterwarnings("ignore", category=DeprecationWarning)
```

```python
df = pd.DataFrame({
    "Transaction ID": [1, 2, 3, 4, 5],
    "Items": [
        "Bread, Milk",
        "Bread, Diaper, Beer, Eggs",
        "Milk, Diaper, Beer, Coke",
        "Bread, Milk, Diaper, Beer",
        "Bread, Milk, Diaper, Coke"
    ]
})
df.to_csv("Groceries_dataset.csv", index=False)
print(df)
```

```python
data = pd.read_csv("Groceries_dataset.csv")

print(data.head())
```

```python
transactions = df["Items"].apply(lambda x: [item.strip() for item in x.split(",")]).tolist()
```

```python
te = TransactionEncoder()
te_array = te.fit(transactions).transform(transactions)
transaction_df = pd.DataFrame(te_array, columns=te.columns_)
```

```python
frequent_itemsets = apriori(
    transaction_df,
    min_support=0.6,
    use_colnames=True
)
```

```python
print("Frequent Itemsets:")
print(frequent_itemsets)
```

```python
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.6
)
```

```python
print("\nAssociation Rules:")
print(rules[[
    "antecedents",
    "consequents",
    "support",
    "confidence",
    "lift"
]])
```

## Dataset2  
_______ Strawberry || Coffee || Orange || Vanilla  
Male   23 || 18 || 8 || 8  
Female 15 || 6 || 12 || 10  
Q1. State null hypothesis and alternate hypothesis  
Q2. Calculate the degree of Freedom  
Q3. Create the expected frequency table for this data.


```python
data = pd.DataFrame(
    [[23, 18, 8, 8],
     [15, 6, 12, 10]],
    index=["Male", "Female"],
    columns=["Strawberry", "Coffee", "Orange", "Vanilla"]
)
```

```python
print("Null hypothesis: There is no association between gender and flavors.")
print("Alternative hypothesis: There is an association between gender and flavors.")
```

```python
print("Observed Frequency Table:")
print(data)
```

```python
chi2, p_value, degrees_of_freedom, expected = chi2_contingency(data)
```

```python
expected_table = pd.DataFrame(
    expected,
    index=data.index,
    columns=data.columns
)
```

```python

print("\nExpected Frequency Table:")
print(expected_table.round(2))
print("\nDegree of Freedom:", degrees_of_freedom)
print("Chi-square value:", round(chi2, 2))
print("P-value:", round(p_value, 4))
```
