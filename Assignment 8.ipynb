# With Dataset
Dataset Source: https://www.kaggle.com/datasets/subhashinimariappan/numerical-dataset

```python
df=pd.read_csv("/content/drive/MyDrive/SEM 5/DMPM LAB/phpB0xrNj.csv")
```

```python
df.head()
```

```python
df.columns
```

```python
numerical_df = df.drop('class', axis=1)
correlation_matrix = numerical_df.corr()
print(correlation_matrix)
```

```python
single_corr = df["f1"].corr(df["f2"])
print(single_corr)
```

```python
sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 6))
```

```python
sns.regplot(
    data=df,
    x="f1",
    y="f2",
    scatter_kws={"alpha": 0.6, "color": "#1f77b4"},
    line_kws={"color": "red", "linewidth": 2}
)
plt.title(f"Scatter Plot of f1 vs f2\n(Pearson Correlation r = {single_corr:.4f})", fontsize=14, fontweight='bold')
plt.xlabel("Feature: f1", fontsize=12)
plt.ylabel("Feature: f2", fontsize=12)
plt.show()
```

```python
print(numerical_df.cov())
```

```python
data = {
    'Hours_Studied': [2, 3, 4, 5, 6],
    'Exam_Score': [50, 58, 65, 70, 85],
    'Sleep_Hours': [8, 7, 7, 6, 5]
}
```

```python
df = pd.DataFrame(data)
```

```python
matrix = df.corr()
print("Correlation Matrix:\n", matrix)
```

```python
single_corr = df['Hours_Studied'].corr(df['Exam_Score'])
print(f"\nCorrelation between Study Hours and Exam Score: {single_corr:.3f}")
```

```python
sns.regplot(
    data=df,
    x="Hours_Studied",
    y="Exam_Score",
    scatter_kws={"alpha": 0.6, "color": "#1f77b4"},
)
plt.title(f"Scatter Plot of f1 vs f2\n(Pearson Correlation r = {single_corr:.4f})", fontsize=14, fontweight='bold')
plt.xlabel("Feature: Hours_Studied", fontsize=12)
plt.ylabel("Feature: Exam_Score", fontsize=12)
plt.show()
```
