# MATPLOTLIB
Matplotlib is an open-source library for creating static, animated and interactive visualizations in Python.
1. It supports line plots, bar charts, histograms, scatter plots and 3D visualizations.
2. Pyplot Module: Provides a procedural, MATLAB-like interface that simplifies plot generation for quick data exploration.
## Pyplot
Pyplot is a module within Matplotlib that provides a MATLAB-like interface for making plots. It simplifies the process of adding plot elements such as lines, images and text to the axes of the current figure.


```python
import matplotlib.pyplot as plt
import pandas as pd

df=pd.read_csv("/content/drive/MyDrive/SEM 5/DMPM LAB/tips.csv")
```

```python
df.columns
```

```python
df.head()
```

# BOX PLOT

```python
data = df['total_bill']
plt.boxplot(data)
plt.xlabel("Groups")
plt.ylabel("Values")
plt.title("Box Plot")
plt.show()
```

# SCATTER PLOT


```python
x=df['tip']
y=df['size']
plt.scatter(x,y,color='r')
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("Scatter Plot")
plt.show()
```

# HISTOGRAM

```python
x=df['size']
plt.hist(x,bins=7,color='black')
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("Histogram")
plt.show()
```

# LINE CHART

```python
df_sorted = df.sort_values('total_bill')
plt.plot(df_sorted['total_bill'], df_sorted['tip'],color="orange")
plt.xlabel("Total Bill")
plt.ylabel("Tips")
plt.title("Total Bill vs Tips")
plt.show()
```

# PIE CHART

```python
data = df['sex'].value_counts()
plt.pie(data, labels=data.index, autopct='%0.2f%%')
plt.title("Customers by Gender")
plt.show()
```

# BAR PLOT

```python
data=df['sex']
value=df['tip']
plt.bar(data,value,color='#003366')
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("Bar Plot")
plt.show()
```

# SEABORN
Seaborn is a Python library for creating statistical visualizations. It provides clean default styles and color palettes, making plots more attractive and easier to read. Built on top of Matplotlib and integrated with pandas data structures, Seaborn makes data visualization easier and more consistent.

```python
import seaborn as sns
```

# BOX PLOT

```python
data = df['total_bill']
sns.boxplot(data)
plt.xlabel("Groups")
plt.ylabel("Values")
plt.title("Box Plot")
plt.show()
```

# SCATTER PLOT

```python
x=df['tip']
y=df['size']
sns.scatterplot(x=x,y=y,color='r')
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("Scatter Plot")
plt.show()
```

# HISTOGRAM

```python
x=df['size']
sns.histplot(x,bins=7,color='brown')
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("Histogram")
plt.show()
```

# LINE CHART

```python
x=df['total_bill']
y=df['tip']
sns.lineplot(x=x,y=y,color='orange')
plt.xlabel("Total bill")
plt.ylabel("Tips")
plt.title("Line Chart")
plt.show()
```

# BAR PLOT

```python
data=df['sex']
value=df['tip']
sns.barplot(x=data,y=value,color='#003366')
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("Bar Plot")
plt.show()
```
