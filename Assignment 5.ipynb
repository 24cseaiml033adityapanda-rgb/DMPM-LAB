```python
# Q1.
import statistics as st
data=[10,20,30,40,50]
mean=st.mean(data)
sd=st.stdev(data)
zscore=[]
for i in data:
  zscore.append((i-mean)/sd)
print("Z-Score: ",zscore)
dataMin=min(data)
dataMax=max(data)
minmax=[]
for i in data:
  minmax.append((i-dataMin)/(dataMax-dataMin))
print("Min-Max: ",minmax)
decimalScaling=[]
for i in data:
  decimalScaling.append(i/10)
print("Decimal Scaling: ",decimalScaling)
```

```python
#Load dataset
import pandas as pd
df=pd.read_csv("/content/drive/MyDrive/SEM 5/DMPM LAB/loanData.csv")
print(df)
```

```python
#Q2
loanAmount=df['loan_amnt']
maxAmount=max(loanAmount)
minAmount=min(loanAmount)
minmax=[]
for i in loanAmount:
  minmax.append((i-minAmount)/(maxAmount-minAmount))
print("Min-Max: ",minmax)
```

```python
#Q3
maxAmount=max(loanAmount)
j=len(str(int(maxAmount)))
decimalScaling=[]
for i in loanAmount:
  decimalScaling.append(i/pow(10,j))
print("Decimal Scaling: ",decimalScaling)
```

```python
#Q4
mean=st.mean(loanAmount)
var=st.variance(loanAmount)
sd=st.stdev(loanAmount)
print("Mean: ",mean)
print("Variance: ",var)
print("Standard Deviation: ",sd)
zscore=[]
for i in loanAmount:
  zscore.append((i-mean)/sd)
print("Z-Score: ",zscore)
```
